# 🎯 XGBoost Regression: Step-by-Step with Tiny Dataset
This notebook walks through how XGBoost performs regression using a simple dataset with only 4 data points.

## 📊 Dataset
We are given the following dataset:

```
import pandas as pd

data = pd.DataFrame({
    'CGPA': [6.7, 9.0, 7.5, 5.0],
    'Package': [4.5, 11.0, 6.0, 8.0]
})
data
```

## 🧠 Step 1: Base Model (Predict Mean)
We start by predicting the mean of the 'Package' column.

```
mean_package = data["Package"].mean()
print("Stage-1 Base Model Prediction (Mean):", mean_package)

data["Stage1_Prediction"] = mean_package
data
```

## ❌ Step 2: Calculate Pseudo Residuals
These are the errors from the base model.

```
data["Residual1"] = data["Package"] - data["Stage1_Prediction"]
data
```

## 🌳 Step 3: Train First Decision Tree on Residuals

```
from sklearn.tree import DecisionTreeRegressor

tree1 = DecisionTreeRegressor(max_depth=2)
tree1.fit(data[["CGPA"]], data["Residual1"])

data["Tree1_Pred"] = tree1.predict(data[["CGPA"]])
data
```

## ➕ Step 4: Apply Learning Rate and Update Predictions

```
learning_rate = 0.3
data["Stage2_Prediction"] = data["Stage1_Prediction"] + learning_rate * data["Tree1_Pred"]
data
```

## 📉 Step 5: New Residuals after Stage 2

```
data["Residual2"] = data["Package"] - data["Stage2_Prediction"]
data
```

## 🌳 Step 6: Train Second Decision Tree on New Residuals

```
tree2 = DecisionTreeRegressor(max_depth=2)
tree2.fit(data[["CGPA"]], data["Residual2"])

data["Tree2_Pred"] = tree2.predict(data[["CGPA"]])
data
```

## ➕ Step 7: Update with Tree2 Output

```
data["Final_Prediction"] = data["Stage2_Prediction"] + learning_rate * data["Tree2_Pred"]
data
```

## 📉 Step 8: Final Residuals

```
data["Final_Residual"] = data["Package"] - data["Final_Prediction"]
data
```

## 📊 Step 9: Visualize Predictions Over Stages

```
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
plt.scatter(data["CGPA"], data["Package"], color='red', label="Actual", s=100)
plt.plot(data["CGPA"], data["Stage1_Prediction"], label="Stage-1 (Mean)", linestyle="--")
plt.plot(data["CGPA"], data["Stage2_Prediction"], label="Stage-2 (After Tree1)", marker="o")
plt.plot(data["CGPA"], data["Final_Prediction"], label="Stage-3 (After Tree2)", marker="D", linewidth=2)
plt.xlabel("CGPA")
plt.ylabel("Package (LPA)")
plt.legend()
plt.grid(True)
plt.title("📈 XGBoost Regression Approximation Curve")
plt.show()
```

## ✅ Final Summary
This notebook demonstrated how XGBoost solves a regression task in stages by correcting residuals.
- Started with mean
- Tree1 learned to predict residuals
- Tree2 further improved accuracy

Each tree attempts to fix what the previous one got wrong.
This is the essence of gradient boosting! 🎉
