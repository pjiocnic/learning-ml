# XGBoost Manual 3-Stage Walkthrough with Multiple Features

This notebook simulates how XGBoost builds a regression model stage-by-stage.

**Scenario**:
- We want to predict a student's **Package (LPA)** based on three features:
  1. CGPA
  2. Marks in 12th
  3. Internship score

**Learning Rate**: 0.3  
**Stages**: We'll construct a 3-stage model manually without using XGBoost.

Each stage attempts to minimize the residual errors from the previous stage using a simple decision stump.


```python
# Step 1: Import necessary libraries
import pandas as pd
```

```python
# Step 2: Create the dataset
data = pd.DataFrame({
    'CGPA': [6.7, 9.0, 7.5, 8.0],
    'Marks_12th': [75, 90, 78, 85],
    'Internship': [8, 9, 7, 10],
    'Package': [4.5, 11.0, 6.0, 8.0]  # in LPA
})
data
```

## Stage 1: Base Prediction using Mean
We predict all outputs as the mean of the actual target values (Package).

```python
# Calculate base prediction (mean of Package)
mean_pred = data['Package'].mean()
data['Stage1_Pred'] = mean_pred
data['Residual_1'] = data['Package'] - data['Stage1_Pred']
data
```

## Stage 2: Tree 1 based on CGPA
We simulate a decision stump:
- If CGPA <= 7.0 → output = -1.5
- Else → output = +1.2

This output is then scaled by learning rate η = 0.3 and added to previous prediction.

```python
# Define learning rate
eta = 0.3

# Simulate Tree 1 Output
data['Tree1_Output'] = data['CGPA'].apply(lambda x: -1.5 if x <= 7.0 else 1.2)
data['Stage2_Pred'] = data['Stage1_Pred'] + eta * data['Tree1_Output']
data['Residual_2'] = data['Package'] - data['Stage2_Pred']
data
```

## Stage 3: Tree 2 based on Marks_12th
Next we train a stump:
- If Marks_12th <= 78 → output = +0.8
- Else → output = -0.5

```python
# Simulate Tree 2 Output
data['Tree2_Output'] = data['Marks_12th'].apply(lambda x: 0.8 if x <= 78 else -0.5)
data['Stage3_Pred'] = data['Stage2_Pred'] + eta * data['Tree2_Output']
data['Residual_3'] = data['Package'] - data['Stage3_Pred']
data
```

## Summary
This manually constructed 3-stage ensemble model shows how boosting incrementally corrects residual errors.
